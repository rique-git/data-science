```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# --- Create the dataframe ---
data = {
    "Joint": ["tornozelo", "joelho", "anca"],
    "Normal": [5, 6, 17],
    "Antes": [-10, 4, 8],
    "Depois": [3, 5, 12]
}

df = pd.DataFrame(data)
df.head()
```

```python
# --- Plot ---
plt.figure()

plt.plot(df["Joint"], df["Normal"], marker='o', color = "green", label="Normal")
plt.plot(df["Joint"], df["Antes"], marker='o', color = "red", label="Antes")
plt.plot(df["Joint"], df["Depois"], marker='o', color = "blue", label="Depois")

plt.xlabel("Articulação")
plt.ylabel("Angulo")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# --- Create the dataframe ---
anca = {
    "Tipo": ["Normal", "Antes", "Depois"],
    "Initial Contact": [7.8, 5, 6],
    "Mid-stance": [-3.7, 4, -1],
    "Terminal Stance": [-15.4, -10, -12],
    "Pré-swing": [-15.2, -9, -13],
    "Toe-off": [-7.5, -3, -6],
    "Mid-swing": [5.7, 10, 5],
    "Terminal Swing": [7.3, 11, 10],
}

df = pd.DataFrame(anca)
df.head()
```

```python
# --- Extract x-axis (all columns except Tipo) ---
x = df.columns[1:]   # gait phases

plt.figure(figsize=(10, 5))

# --- Plot the three conditions ---
plt.plot(x, df.iloc[0, 1:], marker='o', color='green', label='Normal')
plt.plot(x, df.iloc[1, 1:], marker='o', color='blue', label='Antes')
plt.plot(x, df.iloc[2, 1:], marker='o', color='red', label='Depois')

plt.xlabel("Fase da Marcha")
plt.ylabel("Ângulo (graus)")
plt.xticks(rotation=45)
plt.legend()
plt.title("Anca")
plt.tight_layout()

plt.show()
```
